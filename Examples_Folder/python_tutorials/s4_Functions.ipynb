## Functions

Writing a function is the key to modular programming. Functions are a collection of code that together perform a unifying task. Some of the advantages of using functions:
- You can call them in your code whenever you need that function and do not need to copy over lines of code.
- You can use the same function, but apply it to different input variables.
- It increases code readability and allows for easier debugging.

Here's how a function looks:

```
def function-name(arg1, arg2, arg3, ...):
  '''Function description line.'''

  code 1
  code 2
  code 3

  return something/nothing
```

Let's go through the assignment together.

````python
# NOTE: mystery number generation
# the mystery number will be saved as a variable
# called myst_num

from random import randint

myst_num=randint(1,10)
````

Now, let's think about how we can find this number.

We know the number is between [1,10] and we know how to make a loop and how to compare things now. 

Let's give it a go.

````python
for i in range(1,11): #looping over all possibilities
  if i == myst_num: #checking if we found the number
    print("Your number is: ", i)
    break #if we find the number, we don't need to continue the loop
````

Now let's make a function that does exactly this.

````python
def num_finder(input_num):
  '''Takes in an unknown variable and returns its value.'''

  for i in range(1,11): 
    if i == input_num: #checking if we found the number
      print("Your number is: ", i)
      break

  return i #last value of i equals our mystery number
````

Let's check our code. You call a function by simply typing its name and then giving it its arguments.

Since our code is returning a number, we can assign it to a new variable.

````python
my_guess = num_finder(myst_num)
````

````python
print(my_guess)
````

### Getting things to the next level

Now you know how to write a simple function. Note that it is always a good practice to test the code you want to put in a function separately (line-by-line) before putting everything into a function. 

There are multiple ways by which we can improve our function. Let's take a look.

**Using return as an end statement:** Return means the end of function. When you get to the `return` statement, your function will end and loops will exit automatically.

Let's see how our function will look like:

````python
def num_finder_mod(input_num):
  '''Takes in an unknown variable and returns its value.'''

  for i in range(1,11): 
    if i == input_num: #checking if we found the number
      print("Your number is: ", i)
      return i
````

````python
num_finder_mod(myst_num)
````

**More detailed explanation of your function:** We can add more explanation to our function at the beginning. This helps other people who look at your code as well your future self.

Let's take a look. You can see in the code below that we added some more details about what our input and output should be.

````python
def num_finder_mod(input_num):
  '''
  Takes in an unknown variable and returns its value.
  Args:
    input_num: an integer between [1,10]

  Returns:
    An integer that is the calculated value of the input
  '''

  for i in range(1,11): 
    if i == input_num: #checking if we found the number
      print("Your number is: ", i)
      return i
````

````python
num_finder_mod(myst_num)
````

**Testing:** Building tests in the function early in your project helps you debug much more easily in the future. What should we test? The input. We need to make sure the input is actually what the function "expects" it to be, otherwise our function may break or behave in unexpected/undesirable ways.

Let's take a look.

````python
#let's give our function something that is not a number
t1 = 'hi'
num_finder(t1)
````

````python
#now let's try a float
t2 = 2.3
num_finder(t2)
````

You can see that when you give the function the wrong input, it just goes through the loop. The `if statement` is never satisfied and the `for loop` finishes. The function returns the last value of i which is 10.

What if we use our modified function which has the return statement in the loop?

````python
num_finder_mod(t2)
````

````python
my_guess2 = num_finder_mod(t2)
````

````python
my_guess2
````

You can see that nothing happens. There is no error. But, your function doesn't make a guess. This may seem trivial, but things like this can cause huge challenges when you have a big coding project.

Let's see how we can fix it in a simple way.

````python
def num_finder_mod2(input_num):
  '''
  Takes in an unknown variable and returns its value.
  Args:
    input_num: an integer between [1,10]

  Returns:
    An integer that is the calculated value of the input
  '''

  #checking if the input is integer
  if type(input_num) != int:
    print("you need to input an integer number")
    #exiting the function because there is no point in looping
    return #yes, you can return without returning anything!

  #checking if the input is between 1 and 10
  if input_num > 10 or input_num < 1:
    print("your number should be between 1 and 10")
    #exiting the function because there is no point in looping
    return

  for i in range(1,11): 
    if i == input_num: #checking if we found the number
      print("Your number is: ", i)
      return i
````

````python
num_finder_mod2(t1)
````

````python
t3 = -1
num_finder_mod2(t3)
````

### Importing packages and libraries

Functions offer Python modularity. You can call them at any part of the code. You can even write a package that contains many different useful functions that you can use later on.

The **first step** to access a package is to install it on your computer. Depending on how you have python configured, you can use `pip install package-name` or `conda install package-name` or other code to install packages.

### Optional advanced options

**Raising errors:** Our improved function is better. But, as you noticed, there is still no error. Your code runs fine. You just get a prompt that says something is wrong.

Python has a nice way to help you write your own prompt and fail the code if something like this happens. Below, we show you an example for your code. You can learn more about this [here](https://docs.python.org/3/tutorial/errors.html).

````python
def num_finder_adv1(input_num):
  '''
  Takes in an unknown variable and returns its value.
  Args:
    input_num: an integer between [1,10]

  Returns:
    An integer that is the calculated value of the input
  '''

  #checking if the input is integer
  if type(input_num) != int:
    raise ValueError("Oops! You need to enter an integer")

  #checking if the input is between 1 and 10
  if input_num > 10 or input_num < 1:
    raise Exception("your number should be between 1 and 10")

  for i in range(1,11): 
    if i == input_num: #checking if we found the number
      print("Your number is: ", i)
      return i
````

**Typing:** One of the challenges in coding in python is that variables are not assigned types, and you can fluidly assign different types to the same variable.

Using type annotations in functions can help you get a clearer sense of what the inputs and outputs are. It gives a hint about what types are expected. Note that it won't give you an error if the types are wrong. You can read more about typing [here](https://docs.python.org/3/library/typing.html).

Let's take a look.

````python
def num_finder_adv2(input_num: int) -> int:
  '''
  Takes in an unknown variable and returns its value.
  Args:
    input_num: an integer between [1,10]

  Returns:
    An integer that is the calculated value of the input
  '''

  #checking if the input is integer
  if type(input_num) != int:
    raise ValueError("Oops! You need to enter an integer")

  #checking if the input is between 1 and 10
  if input_num > 10 or input_num < 1:
    raise Exception("your number should be between 1 and 10")

  for i in range(1,11): 
    if i == input_num: #checking if we found the number
      print("Your number is: ", i)
      return i
````
