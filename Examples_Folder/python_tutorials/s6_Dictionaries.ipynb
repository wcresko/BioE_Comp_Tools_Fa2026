## Dictionaries

Dictionaries are popular methods for data storage in python. Data in dictionaries are stored in Key:Value format. Check below:

```
D = {
  key1: value1,
  key2: value2,
  key3: value3,
}
```

The values in the dictionary can be of type int, str, list, or even another dictionary. They don't need to be of the same type or length either. The keys in dictionary can also be of mixed type. For example, you can use a mixture of numbers and strings. However, they must be unique. You can access all the keys using this command:

```
len(D) #gives you the number of keys
D.keys() #gives you all the keys
```

`D.keys()` generates a variable of type dict_keys. You can easily get the list of keys by typing `list(D.keys())`. Similarly, you can get all the values in a dictionary using `D.values()`. `D.items()` give you pairs of [(key1, value1),(key2, value2), ...].

To access the value of a given key, you can simply use:

```
D[key1]
```

You can loop through all the keys in a list using:

```
for key in D.keys():
  item = D[key]
  other action
```

Adding a new key to the dictionary is very easy. All you need to do is:

```
D[new_key] = new_value
```

Let's try to write down the code that answers the exercise. We will first do it the difficult way and then we use an easier way.

````python
#Creating a dictionary of salaries of 10 employees/month
salary_dict = {
    "emp1": 2000,
    "emp2": 2500,
    "emp3": 2200,
    "emp4": 2000,
    "emp5": 3000,
    "emp6": 2700,
    "emp7": 3500,
    "emp8": 3000,
    "emp9": 3200,
    "emp10": 2000,
}
````

````python
#let's check the items in the dictionary
salary_dict.items()
````

You can see that the items are tuples of size two. The first element in the tuple is the key and the second one is our value.

````python
#let's check the length of the dictionary
len(salary_dict)
````

````python
#let's loop through all the keys
for key in salary_dict.keys():
  print(key)
````

In order to find the key for the lowest value, we can generate a place-holder variable for the key, for example `min_key` and a place holder variable for the minimum value, for example `min_val`. We can then loop through all the items and if the value is less than min_value, re-assign our `min_key` to the new key. Let's see how it works.

````python
#creating the place holder variable to store min key
min_key = "" #empty string

#creating a place holder variable to store min value
#we use an unrealistically large number to make sure
#our values will be smaller than this
min_val = 10000000 

#looping through items
for item in salary_dict.items():
  #comparing value with min_val
  #remember that item is a tuple (key, val)
  #so value is item[1]
  if item[1] < min_val:
    #if smaller, we want minimum value to be changed to this number
    min_val = item[1]
    #and key to change to this key
    min_key = item[0]

print("minimum value is ", min_val,
      "and its key is ", min_key)
````

What if we didn't have a guess of the reasonable range of values?

In that case, you can assign the variables to be the first item in the loop and then move to the loop from there. An example is shown below.

````python
#setting a counter
counter = 0

#looping through items
for item in salary_dict.items():
  #if we're at the first item
  if counter == 0:
    min_val = item[1]
    min_key = item[0]
    #making sure we change the counter
    counter += 1
  #comparing value with min_val
  #remember that item is a tuple (key, val)
  #so value is item[1]
  if item[1] < min_val:
    #if smaller, we want minimum value to be changed to this number
    min_val = item[1]
    #and key to change to this key
    min_key = item[0]

print("minimum value is ", min_val,
      "and its key is ", min_key)
````

Now let's take a look at a faster way to run the same code.

**Step 1.** Making a list of all values.

````python
salary_list = list(salary_dict.values())
print(salary_list)
````

**Step 2.** Finding the index of the minimum value in the list (check *s5_Lists* if you need a refresher on this.

````python
min_val = min(salary_list)
min_idx =  salary_list.index(min_val)

print(min_val, min_idx)
````

**Step 3.** Getting the key

````python
min_key = list(salary_dict.keys())[min_idx]

print(min_key)
````

### Saving a dictionary

<font color='grey' size='2'> These examples are copied from [Python Tutorials](https://pythonspot.com/save-a-dictionary-to-a-file/)

The two most common formats of saving dictionaries are **csv**, **json**, **text** files, and **pickle files**. Let's see how each work.

#### csv format

You need to use csv module to be able to use this. Then simply use a for loop to 

```
# load csv module
import csv

# define a dictionary with key value pairs
dict = {'Python' : '.py', 'C++' : '.cpp', 'Java' : '.java'}

# open file for writing, "w" is writing
w = csv.writer(open("output.csv", "w"))

# loop over dictionary keys and values
for key, val in dict.items():

    # write every key and value to file
    w.writerow([key, val])
```

#### json file

Today, a JSON file has become more and more common to transfer data in the world. JSON (JavaScript Object Notation) files are lightweight data-interchange format that are becoming increasingly popular.

JSON is a text format that is completely language independent but uses conventions that are familiar to programmers of the C-family of languages, including C, C++, C#, Java, JavaScript, Perl, Python, and many others.

Here's how you can save your dictionary in a json format:

```
# load json module
import json

# python dictionary with key value pairs
dict = {'Python' : '.py', 'C++' : '.cpp', 'Java' : '.java'}

# create json object from dictionary
json = json.dumps(dict)

# open file for writing, "w" 
f = open("dict.json","w")

# write json object to file
f.write(json)

# close file
f.close()
```

If you open the file, it will look like this:

```
{"Python": ".py", "C++": ".cpp", "Java": ".java"}
```

#### Text file

Although less common, you can save your dictionary as a text file using the code below:

```
# define dict
dict = {'Python' : '.py', 'C++' : '.cpp', 'Java' : '.java'}

# open file for writing
f = open("dict.txt","w")

# write file
f.write( str(dict) )

# close file
f.close()
```

Note that this code saves everything as strings.

#### Pickle file

In Python, pickle is a built-in module that implements object serialization. It is both cross-platform and cross language, meaning that it can save and load objects between Python programs running on different operating systems, as well as between Python running on different platforms.

Here's how you can save your dictionary as a pickle file:

```
# load pickle module
import pickle

# define dictionary
dict = {'Python' : '.py', 'C++' : '.cpp', 'Java' : '.java'}

# create a binary pickle file 
f = open("file.pkl","wb")

# write the python object (dict) to pickle file
pickle.dump(dict,f)

# close file
f.close()
```
