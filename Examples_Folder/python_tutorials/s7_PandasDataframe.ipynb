## Pandas Dataframe

Dataframes provide a convenient way to store and handle data. They have many built-in functions that you can use for understanding and manipulating data. They even provide some easy plotting options. 

Let's take a look.

### The basics

````python
#first, we need to import pandas
import pandas as pd
````

````python
#Let's create a dataframe 
#As you can see, the input is a dictionary
#One big difference is that to make a dataframe
#you need to have all values be of same length!
#we'll see why

df = pd.DataFrame(
    {
      "score": [1.132,6.816,159.113,70.82,94.979,
                101.4,11.33,39.817,24.846,48.497],
      "hbond_count": [2,3,4,3,2,3,4,2,2,2],
      "cis_bonds" : [0,0,1,0,0,1,0,0,0,1],
      "name": ["result_0001","result_0003","result_0004",
               "result_0006","result_0007","result_0013",
               "result_0015","result_0016","result_0025",
               "result_0029"],
  }
)
````

````python
#Let's take a look at our dataframe
df
````

You can see that the keys in our dictionary are now the column names of our dataframe and the items are rows. There is also an index row that is added.

When we have lots of data, it's better to just take a look at a few rows in our dataframe. in this case, you can use the function `head`. Let's see how it works.

````python
df.head(5)
# 5 is actually the default. You can change it to any number
````

````python
#To see the number of rows in dictionary
len(df)
````

We can choose to look at only one of the columns. 

````python
df['score']
````

Or even a specific value in a column.

````python
df['score'][0]
````

Note that this works like a list but is not really a list. The type of this is a pandas series.

````python
type(df['score'])
````

If you need to, you can change this to a list.

````python
list(df['score'])
````

### Accessing dataframe values

Let's go through different ways we can make selections in our dataframe and look at data.

````python
#looking at first 5 rows
df.head()
````

#### Selecting by labels

`loc` makes selections by labels.

````python
#selecting by index
df.loc[2]
````

````python
#selecting all rows for a given column
df.loc[:, ["score"]]
````

````python
#selecting some rows and some columns
df.loc[1:2,["score"]]
````

#### Selecting by position

`iloc` makes selections by position. Let's take a look.

````python
#index position 1
df.iloc[1] #same as df.iloc[1,:]
````

````python
#selecting certain rows and column
df.iloc[1:3, 2:4]
````

#### Boolean indexing

We can also look at rows or columns based on their values or a condition. Let's take a look.

````python
#condition
#all rows where score is less than 10
df[df["score"] < 10]
````

````python
#checking to find locations where elements are in a list
df[df["name"].isin(["result_0001", "result_00025", "result_0016"])]
````

### Statistical analysis on data

Pandas provide a simple way to get major statistical information on the data without you needing to calculate it. Let's check that out below.

````python
#getting mean value of data
df.mean()
````

You can see that this is only defined for numerical values, so it is ignoring the name column in our dataframe.

````python
#getting the standard deviation of columns
df.std()
````

You can get a number of useful statistical measurements using `describe` function. Check below:

````python
#Getting statiscical analysis on a column
df['score'].describe()
````

Pretty neat, right? 

Let's see what happens if you give it a column that does not have numerical values.

````python
df['name'].describe()
````

Let's run it on the whole dataframe.

````python
df.describe()
````

You see that our dataframe removed the column with non-numerical values. 

You can also define your own functions to get information about the columns. Dataframes provide a powerful tool to that using `apply`. The function can be written prior or can be defined in-line (`lambda`). Let's take a look at two examples:

````python
#using a pre-defined function from numpy
import numpy as np

df.apply(np.cumsum) #calculates cumulative sum
````

Note that you do not need to add `()` after the function inside apply. The function takes on values of each [row,column] and gives it as an input to your function.

````python
#defining function in line
#x here is the value at each location
df.apply(lambda x: x*2)
````

### Modifying your dataframe

Let's see how we can modify our dataframe.

#### Making a dataframe from another one

You can copy an entire dataframe or select some of its rows or columns to make a new dataframe.



````python
#copying
df_c = df.copy()
len(df_c.index)
````

````python
#selecting certain columns
df_col = df[["score", "hbond_count"]]
len(df_col.columns)
````

````python
#selecting certain rows and columns
df_sel = df.iloc[1:4, 2:3]
df_sel
````

#### Adding a new column
Adding a new column is as easy as this:

```
df['new_column`] = [item1, item2,...]
```

Let's try adding a new column that shows energy/hydrogen bond in our dataframe.

````python
df["score_per_hbond"] = df["score"]/df["hbond_count"]
df.head()
````

#### Adding a new row

You can add a new row by explicitly giving its idex location using loc. You need to make sure the number of elements you give matches the total number of columns.

````python
#adding a row in the last index
df.loc[len(df.index)] = [2.5, 2 , 0, "new_row", 1.25]
len(df)
````

````python
#let's take a look at last 5 rows. You can see the new added row here.
df.tail()
````

#### Dropping a column or row

You can drop a column or a row in a dataframe using `drop`.

````python
#let's first make a copy of our dataframe
df_copy = df.copy()
df_copy
````

Now let's drop one of the columns.

````python
df_copy.drop(columns=["score_per_hbond"])
````

You can also drop that column by typing:

```
df_copy.drop(["score_per_hbond"], axis=1)
```

axis=1 means you're dropping a column.

Index dropping is very similar. Let's try:

````python
df_copy.drop(index=[10])
# you could also do this using
#df_copy.drop([10])
````

### Saving dataframes and reading them from file

#### Saving dataframes

There are multiple ways to store your dataframe. 

**csv:** The most common format for storing dataframes is a csv format. Pandas dataframes have a built-in function that allows you to do so. Here's how:

```
df.to_csv("df_name.csv")
```

If you saved your dataframe as a csv, here's how you can read it later;

```
pd.read_csv("name.csv", sep="\t")
```

In the code above, we are reading a csv file that uses tab `\t` as the field separator. You can use other separators depending on your csv file. For a complex file, you can use `"\s+"`. Read more about read_csv [here](https://pandas.pydata.org/docs/reference/api/pandas.read_csv.html).

**pickle file:** Another way to store your dataframe is as a pickle file. Pickle file is another data structure in python. You can read more about it [here](https://www.geeksforgeeks.org/understanding-python-pickling-example/). 

Here's how you save as a pickle file:

```
df.to_pickle("my_data.pkl")
```

Reading is also simple:

```
df = pd.read_pickle("my_data.pkl")
```
