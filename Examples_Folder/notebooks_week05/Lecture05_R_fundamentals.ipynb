## 1 · Arithmetic

R follows the usual order of operations. The exponent is `^` (in Python it will be `**`).

````r
4 * 4
````

````r
(4 + 3 * 2^2)      # multiplication and ^ before +
````

````r
7 %/% 2            # integer division
7 %% 2             # remainder (modulo)
````

## 2 · Comments, names and style

Comments run from `#` to the end of the line. R is **case sensitive**, and assigning to an existing name silently overwrites it.

````r
# This whole line is a comment
n_fish <- 80        # so is everything after the code
n_fish

# Case matters: n_fish and N_fish would be two different variables
````

## 4 · Functions on numbers

`log()` is a built-in function: the thing it operates on goes in parentheses. Results can be assigned and printed.

````r
x + 2
x^2
log(x)             # natural log
````

````r
y <- 67
print(y)

x <- 124
z <- (x * y)^2
print(z)
````

## 5 · Comparisons and logical values

Comparisons return a **logical**: `TRUE`, `FALSE` (or `NA`). Combine them with `&` (and), `|` (or), `!` (not). These power indexing and filtering later in the lecture.

````r
x <- 5
x > 3
x == 10            # equality is ==  (a single = would try to assign!)
x != 10
(x > 1) & (x < 10)
(x < 1) | (x > 4)
!(x > 3)
````

## 6 · Strings

Characters go in quotes. `c()` concatenates values into a vector. Re-using a name overwrites the old value.

````r
x <- "I Love"
print(x)
y <- "Bioengineering"
print(y)
z <- c(x, y)
print(z)
````

## 7 · Working with strings

````r
gene <- "tp53"
nchar(gene)                      # how many characters?
toupper(gene)                    # change case (tolower() goes the other way)
paste(gene, "is my favorite")    # glue with a space
paste0(gene, "_counts.tsv")      # glue with nothing — great for file names
substr(gene, 1, 2)               # characters 1 through 2
sprintf("GC content of %s = %.1f%%", gene, 53.81)   # formatted output
````

## 8 · Vectors

R *thinks* in vectors: `x^2` squares every element at once. (In Python this will need a numpy array.)

````r
x <- c(2, 3, 4, 2, 1, 2, 4, 5, 10, 8, 9)
print(x)
y <- x^2           # element-wise: every value squared
print(y)
````

````r
plot(x, y)         # each value of x against its square
````

# Part 2 · Types of data

Every vector has exactly **one** type. Knowing which one is half of debugging.

| What it holds | R (tibble code) | Example |
|:---|:---|:---|
| TRUE / FALSE | logical (`lgl`) | `TRUE` |
| Whole numbers | integer (`int`) | `3L` |
| Real numbers | double (`dbl`) | `3.0`, `2.5e-3` |
| Text | character (`chr`) | `"low"` |
| Categories | factor (`fctr`) | `factor(c("low","high"))` |
| Dates / date-times | `date` / `dttm` | `as.Date("2026-11-09")` |

Numbers are **doubles by default** (write `3L` to force an integer). Integer division and remainder are `%/%` and `%%`.

## 13 · Checking what type you have

````r
x <- 3.14
class(x)              # what kind of object is this?
typeof(x)             # the low-level storage type
is.numeric(x)
is.character(x)
str(c(a = 1, b = 2))  # compact structure summary — great for big objects
````

````r
class(3L)             # a 'forced' integer
class("3")            # the quotes make it text!
class(TRUE)
class(factor(c("low", "high")))
````

## 14 · Converting between types

R fails *quietly*: the unconvertible becomes `NA` with a warning, and the script keeps going. (Python will fail *loudly* instead.)

````r
as.numeric("3.14")    # string to number
as.character(3.14)    # number to string
as.integer(3.99)      # truncates toward zero — does NOT round
as.numeric("abc")     # NA plus a warning — check for NA after converting!
````

## 15 · One type per vector: coercion

Mix types in `c()` and R silently coerces everything to the most flexible type: logical → integer → double → character.

````r
c(1, 2, 3)            # double
c(1, 2, "three")      # one string turns ALL of them into strings
c(1, 2, TRUE)         # TRUE quietly becomes 1
sum(c(TRUE, TRUE, FALSE, TRUE))   # ... which is genuinely useful: counting!
````

## 16 · Special values: NA, NaN, Inf, NULL

````r
x <- c(1, 2, NA, 4)
is.na(x)                # never test with == NA (the answer would be NA!)
mean(x)                 # one NA poisons the result ...
mean(x, na.rm = TRUE)   # ... unless you ask to remove it
````

````r
0 / 0                  # NaN — not a number
1 / 0                  # Inf
-1 / 0                 # -Inf
is.nan(0 / 0)
is.infinite(1 / 0)
````

## 17 · Beyond vectors: lists

A vector holds many values of **one** type. A **list** holds *anything*, with a name for each element — and a data frame is secretly a list of equal-length vectors (one per column).

````r
fish <- list(id = "AC1", mass_g = 1.42, counts = c(16, 15, 71))
fish$mass_g            # elements by name with $ ...
fish[["counts"]]       # ... or with [[ ]]
str(fish)              # names, types and sizes at a glance
````
