## 1 · Arithmetic

Python follows the usual order of operations. The exponent is `**` (in R it is `^` — in Python `^` is bitwise XOR, a classic trap!).

```python
4 * 4
```

```python
(4 + 3 * 2**2)     # multiplication and ** before +
```

```python
print(7 / 2)       # true division: ALWAYS a float
print(7 // 2)      # integer division
print(7 % 2)       # remainder (modulo)
```

```python
# This whole line is a comment
n_fish = 80         # so is everything after the code
n_fish

# Case matters: n_fish and N_fish would be two different variables
```

## 3 · Assigning variables

Assign with `=`. Names must not start with a digit (they may start with `_`). Python can assign several at once.

```python
x = 2
print(x * 3)
y = x * 3
print(y - 2)

a, b = 1, 2        # multiple assignment in one line
print(a, b)
```

## 5 · Comparisons and logical values

Comparisons return a **bool**: `True` or `False` (capitalized only on the first letter). Combine them with the *words* `and`, `or`, `not`. These power indexing and filtering later in the lecture.

```python
x = 5
print(x > 3)
print(x == 10)         # equality is ==  (a single = would try to assign!)
print(x != 10)
print((x > 1) and (x < 10))
print((x < 1) or (x > 4))
print(not (x > 3))
```

## 6 · Strings and lists

Characters go in quotes. Square brackets `[ ]` build a **list** (R's `c()` builds a vector). `+` joins strings.

```python
x = "I Love"
print(x)
y = "Bioengineering"
print(y)
z = [x, y]             # a list of two strings
print(z)
print(x + " " + y)     # + joins strings
```

## 7 · Working with strings

Python strings are *sequences*: you can slice them directly. **f-strings** (`f"..."`) are the modern way to format output.

```python
gene = "tp53"
print(len(gene))                   # how many characters?
print(gene.upper())                # change case (.lower() goes the other way)
print(gene + " is my favorite")    # glue with +
print(f"{gene}_counts.tsv")        # an f-string builds file names too
print(gene[0:2])                   # characters 1-2 (0-based, stop excluded!)
print(f"GC content of {gene} = {53.81:.1f}%")   # formatted output
```

## 8 · Lists and numpy arrays

A plain list has no vector math — `x**2` on a list is an error. A **numpy array** behaves like an R vector: element-wise everything.

```python
x = [2, 3, 4, 2, 1, 2, 4, 5, 10, 8, 9]    # a list
print(x)
# x**2 would be a TypeError — try it!

[v**2 for v in x]                          # a 'list comprehension' loop
```

```python
import numpy as np

x = np.array([2, 3, 4, 2, 1, 2, 4, 5, 10, 8, 9])   # a numpy array
y = x**2                                 # element-wise, like R
print(x)
print(y)
```

```python
import matplotlib.pyplot as plt

plt.scatter(x, y)                        # each value of x against its square
plt.xlabel("x")
plt.ylabel("y")
plt.show()
```

## 9 · Basic statistics on an array

Python functions come in two flavours: `np.median(x)` (a function) and `x.mean()` (a **method** attached to the array). Try each — and find one I left out (standard deviation?).

```python
print(x.mean())                    # or np.mean(x)
print(np.median(x))
print(x.var(ddof=1))               # sample variance, like R's var()
print(np.log(x))                   # natural log, element-wise
print(np.sqrt(x))
print(x.sum())                     # sum(x) also works
print(len(x))                      # length(x) in R

rng = np.random.default_rng(2026)  # a seeded random generator (like set.seed in R)
print(rng.choice(x, size=len(x), replace=True))   # resample WITH replacement
```

## 10 · Creating arrays with generating functions

Vary the arguments until you can predict the output before running. Remember: `range(10)` starts at 0 and **excludes** the stop.

```python
seq_1 = np.linspace(0.0, 10.0, 101)    # 101 points from 0 to 10 (step 0.1)
print(seq_1)
seq_2 = np.linspace(10.0, 0.0, 101)

seq_square     = seq_2 * seq_2         # element-wise ...
seq_square_new = seq_2**2              # ... same result
print(seq_square_new)
```

```python
print(list(range(10)))             # integers 0..9 (start defaults to 0; stop excluded!)
print(np.arange(0, 10, 0.5))       # like seq(0, 10, by = 0.5) — stop still excluded
print(np.tile(["A", "B"], 3))      # like rep(..., times = 3)
print(np.repeat(["A", "B"], 3))    # like rep(..., each = 3)
```

# Part 2 · Types of data

Every value has exactly one type — and a numpy array or pandas column has one **dtype** shared by all its values, just like an R vector.

| What it holds | Python | R equivalent | Example |
|:---|:---|:---|:---|
| True / False | `bool` | logical (`lgl`) | `True` |
| Whole numbers | `int` | integer (`int`) | `3` |
| Real numbers | `float` | double (`dbl`) | `3.0`, `2.5e-3` |
| Text | `str` | character (`chr`) | `"low"` |
| Categories | `pandas.Categorical` | factor (`fctr`) | `pd.Categorical(["low","high"])` |
| Dates / date-times | `datetime.date` / `datetime.datetime` | `date` / `dttm` | `datetime.date(2026, 11, 9)` |

Python keeps `int` and `float` strictly apart: `7 / 2` is `3.5` (always a float), `7 // 2` is `3`, `7 % 2` is `1`.

## 13 · Checking what type you have

```python
x = 3.14
print(type(x))                 # what kind of object is this?
print(isinstance(x, float))    # is it a float?

arr = np.array([1, 2, 3])
print(arr.dtype)               # ONE dtype for every value in the array
```

```python
print(type(3))
print(type("3"))               # the quotes make it text!
print(type(True))
import pandas as pd
print(pd.Categorical(["low", "high", "low"]))
```

## 14 · Converting between types

Python fails *loudly*: an impossible conversion raises an error and stops the script. (R fails quietly with `NA` instead.)

```python
print(float("3.14"))    # string to number
print(str(3.14))        # number to string
print(int(3.99))        # truncates toward zero — does NOT round
```

```python
# An impossible conversion STOPS the program with a ValueError:
try:
    float("abc")
except ValueError as err:
    print("ValueError:", err)
# (try/except catches the error so the notebook keeps running — more in L07)
```

## 15 · One container, one type? Coercion

A plain list happily mixes types. A numpy array coerces to one type, exactly like an R vector — and that is where reading messy data bites.

```python
print([1, 2, "three"])             # a list happily mixes types — no coercion
print(np.array([1, 2, "three"]))   # a numpy array coerces: ALL strings now
print(sum([True, True, False, True]))   # True counts as 1 — useful: counting!
```

## 16 · Special values: None, nan, inf

```python
x = pd.Series([1, 2, np.nan, 4])
print(x.isna())             # like is.na() — never test with == nan
print(x.mean())             # pandas SKIPS missing values by default!
print(x.mean(skipna=False)) # ask for R-like behavior (NA poisons the mean)
```

```python
print(np.float64(1) / 0)    # inf (numpy warns; plain Python 1/0 raises an error)
print(np.float64(0) / 0)    # nan
print(math.inf, -math.inf)
print(float("nan") == float("nan"))   # False! nan is not even equal to itself
```

## 17 · Beyond vectors: dictionaries

A list or array holds a sequence; a **dictionary** holds *anything*, with a name (key) for each element — and pandas builds a `DataFrame` from a dictionary of equal-length columns the same way R builds a data frame from a list. (Python also has **tuples** `(1, 2)` and **sets** `{1, 2}` — Parisa's s5 and s6 notebooks work through all four.)

```python
fish = {"id": "AC1", "mass_g": 1.42, "counts": [16, 15, 71]}
print(fish["mass_g"])      # elements by key
print(fish.keys())
print(fish["counts"][0])   # containers nest: the list inside the dict
```
