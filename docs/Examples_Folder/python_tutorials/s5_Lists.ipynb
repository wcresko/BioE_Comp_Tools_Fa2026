## Lists

Lists are one of the commonly used **Data Structures** in python. Lists are simply a collection of objects of any type. A list is created:

```
L = [item1, item2, item3, item4]
```

Each item has an index value associated with it that you can access it with. Note that numbering in python starts from 0:

```
L[0] --> item1
L[1] --> item2
```

Because your list is ordered, you can also access more than one element of your list. For example:

```
L[1:3] --> item2, item3 #includes the first index but not the last
L[:3] --> item1, item2, item3 #starts at 0
L[1:] --> item2, item3, item4 #ends at the last element
L[-1] --> item4 #last item
L[-3:-1] --> item3, item4 #includes the last index not the first
```

You can assign values to list locations the same way you assign variables to values:

```
L[0] = new_item
```

Be careful as this will change your list!

You can add to the end of a list using `append`:

```
L.append(item5)
```

Will give you a list of `[item1, item2, item3, item4, item5]`.

You can get the length of a list by typing:

```
len(L)
```

You can add two lists together and create a list that has components of both lists:

```
L1 = [item1, item2]
L2 = [item3, item4]
L3 = L1 + L2  --> [item1, item2, item3, item4]
```

You can read all the elements in the list using a for loop as follows (notice that lists are iterable!):

```
for item in list:
  some code
```

In this way, you have easy access to all the items. However, getting indices can be hard this way. Another way to loop through a list is using the index values. We know indices start at `0` and end at `length - 1`.

```
for i in range(len(list)):
  some code
```
You now have access to indices, and you can get the items by using `list[i]`. Note that I didn't specify 0 at the beginning of my range. By default, python starts the range from 0.

Alternatively:
```
for idx, item in enumerate(list):  # idx will contain the index and item the item in the list at that index
  some code
```

## Step 1. Finding min and max

````python
# first let's assume the first element of the list is the min and max
# just for simplicity. We will fix it later.
min_id = 0
max_id = 0
min_val = test_ar[0]
max_val = test_ar[0]

#loop through our array
#Q: Do we need to start from 0?
for i in range(len(test_ar)):
  #checking if we find a smaller value
  if test_ar[i] < min_val:
    #adjusting our min value and id accordingly
    min_val = test_ar[i]
    min_id = i
  #checking if we find a bigger value
  if test_ar[i] > max_val:
    #adjusting our min value and id accordingly
    max_val = test_ar[i]
    max_id = i

print("The min value is: ", min_val,
      "and it's in location: ", min_id)

print("The max value is: ", max_val,
      "and it's in location: ", max_id)
````

Now let's go to the more challenging task.

## Step 2. Finding the median

We know that to find the median, we need to sort our list. Here's one way you can sort a list. What other ways you can think of?

````python
#generating a list of 0s with the same length as our list
#this list will be the one we sort
sorted_ar = []
for item in test_ar:
  sorted_ar.append(0)

#checking if we did this right
sorted_ar
````

````python
#Now let's loop over elements of our test list, compare them
#and find their new (sorted) locations/indices
#we will be using nested loops for this

#outer loop to get the elements we are checking
for i, curr_val in enumerate(test_ar):
  #initiating the idx - this is the location we will store the index
  #of where the current value should be located in our sorted list
  idx = 0

  #second loop compares current value to all other list elements
  for test_val in test_ar:
    #index location goes up as you find something that is smaller than your current number
    if curr_val > test_val:
      idx += 1
  #saving each value to the correct (sorted) index position in the sorted_list 
  #AND keeping a record of the index in the original list
  sorted_ar[idx] = [curr_val,i]

#let's check our code
sorted_ar
````

## Optional: Other list-like data structures

### Tuples

```
T = (item1, item2, item3)
```
Tuples are another method for storing data similar to lists. Like lists, tuples can have repetitive values and are ordered. Unlike lists, tuples cannot be changed. Therefore, they are very useful to store data that you don't wish to be changed during your code. You can read more about tuples [here](https://www.tutorialspoint.com/python/python_tuples.htm).

### Sets

```
S = {item1, item2, item3}
```
Sets are very similar to lists in that they store values. However, unlike lists and tuples, sets cannot take on repetition and are not ordered. Thus, a very easy way to remove duplicates from a list is to change it to a set: `L = list, S = set(L)`.

Sets are very useful for performing common math operations such as union, intersection, and difference. You can read more about sets [here](https://www.datacamp.com/tutorial/sets-in-python).
