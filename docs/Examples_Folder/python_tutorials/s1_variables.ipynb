## Python variables

Variables are objects in python that hold some value for us. Python allows us to manipulate and combine variables. Like other programming languages, python has two major types of primitive variables: String variables and numerical variables. Numerical variables can be further divided based on their type: integer, float, etc.

Let's see how python handles variables.

But first, let's introduce a very useful function, called `print`. `print` does exactly what its name suggests: it prints whatever you give it. This is extremely helpfully to see the output of your work.

```python
print("hello. This is my first code!")
```

```python
print(2)
```

### Assigning variables

Assigning variables in python is very simple. You just use **=** to assign something to your variable. 

For naming the variables, you can use combinations of letters and numbers and underline. Don't start your variable name with a number and don't use special characters other than underline (so x1 or x_1 is fine, but 1x or x% are not).

```python
# let's assign a string
# in python strings are shown within "" marks
my_str="hello"
print(my_str)
```

```python
my_str="second effort"
print(my_str)
```

You can change the types of variables in python. Let's take a look at some examples below.

```python
#changing int to float
x = 2
print(type(x))

y = float(x)
print(y, type(y))
```

```python
#changing float to int
x = 1.5
print(type(x))

y = int(x)
print(y, type(y)) 
```

QUESTION: Try changing other float variables to int to see how python rounds things up.

```python
#changing numbers to strings
x = 2
print(type(x))

y = str(x)
print(y, type(y))
```

```python
#changing string to int
x = "2"
print(type(x))

y = int(x)
print(y, type(y))
```

**Variable overwrite** As you saw above, python overwrites variables! This can be a blessing and a curse. Make sure you don't accidentally wipe down a variable.

```python
#assigning numbers
my_num=2
print(my_num)
```

```python
my_num="strange!"
print(my_num)
```

As you can see, variable overwrite is independent of the type. This is unlike a language like C++ where you could overload the same variable name but assign it different types. Python simply forgets what the variable contained and remembers the new assignment.

### Simple variable manipulation

Python is very powerful and flexible at variable handling and manipulation. Let's check out some of its functionalities.

For a full list of operators in python, check [this link](https://www.w3schools.com/python/python_operators.asp).

```python
# mathematical calcualtions on numbers
num1=2
num2=13
print(num1+num2)
```

```python
print(2*num1+num2-3)
```

```python
print(num1**3)
```

```python
# two types of division
print(num2/2)
```

```python
print(num2//2)
```

```python
# getting modulus
print(num2%2)
```

```python
# recursive assignment
num1 += 1 #same thing as num1=num1+1
print(num1)
```

You can also perform comparisons in python. Notice that these give us a third type of python primitive: a boolean (aka bool). Booleans are `True` or `False`.

```python
print( num1<num2)
```

```python
#using and, or, not to combine logical statements
# below, first statement is incorrect (False) and second one is correct (True)
# so "and" is incorrect (False)
print(num1>num2 and 3<4) 
```

```python
# whereas with "or", it's enough that one is correct (True)
print(num1>num2 or 3<4) 
```

### Changing variable types

In python you can change the type of a numerical variable to string. This is called "casting" and can be helpful if you want to use a numerical variable in the name of something. To do so, we use the function `str`. 

Let's try this with our summation example above.

```python
print(str1+str(num1))
```
