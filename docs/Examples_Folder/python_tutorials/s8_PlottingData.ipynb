Plots are ways to visualize your data in a concise and meaningful way. Depending on the type of data you have, it may make sense to use different types of plots such as scatter plot, bar plot, whisker plot, violin plot, etc.

### A few simple cases

Let's first check out some simple plots using a set of pre-generated datapoints. 

````python
#importing matplotlib
import matplotlib.pyplot as plt
````

````python
#plotting line
#step 1. generating a sin wave
import numpy as np
x = np.arange(0, 4*np.pi, 0.1)
y = np.sin(x)

#plotting
plt.plot(x, y)
plt.xlabel('x')
plt.ylabel('y')
plt.title('A sine wave')
````

````python
#plotting two line plots
#generating data points
x = np.arange(0, 4*np.pi, 0.1)
y1 = np.sin(x)
y2 = np.cos(x)

#plotting
plt.plot(x, y1)
plt.plot(x, y2)
plt.xlabel('x')
plt.ylabel('y1, y2')
plt.title('Sine and cosine')
plt.legend(['Sine', 'Cosine'])
````

````python
#plotting as points instead of lines
#generating data
x = np.arange(0, 4*np.pi, 0.1)
y = np.sin(x)

#plotting
plt.plot(x, y, 'o')  # also try replacing 'o' by 's' or '.' or '--'
plt.xlabel('x')
plt.ylabel('y')
plt.title('A sine wave')
````

#### Adding subplots

We can make subplots within the same plot. In the following examples, we'll illustrate two different distributions of random numbers in two different subplots. We'll also increase the number of bins in the histogram plots to show the distributions in more detail.

````python
# Generate some random data:
x1 = np.random.rand(10000)
x2 = np.random.randn(10000)

# The first subplot:
plt.subplot(121) # the three numbers refer to the number of rows, number of columns, and which subplot we're making
plt.hist(x1, bins=30)
plt.title('Uniform')
plt.xlabel('x')
plt.ylabel('Count')

# The second subplot:
plt.subplot(122)
plt.hist(x2, bins=30)
plt.title('Normal')
plt.xlabel('x')

# Pro tip: The following line makes everything fit together more nicely. 
# Try uncommenting it and compare the difference:
# plt.tight_layout()
````

#### Adding error bars

When we are working with actual data, we will often want to average it (e.g. averaging over trials in a neuroscience experiment) and include error bars to illustrate how spread out the data is. Let's create some fake data in which the underlying signal is a sine wave, and each trial has noise added to it. Then we'll plot the trial-averaged data together with error bars used to denote the standard deviation. This example will also include some of what we've learned about using for loops and performing operations over arrays.

````python
# Generate some fake data:
t = np.arange(0, 2*np.pi, 0.2)
signal = np.sin(t)
n_timesteps = len(t)
n_trials = 20
y = np.zeros((n_trials, n_timesteps))

# Add noise to the signal:
for i in range(n_trials):
    y[i,:] = signal + 0.2*np.random.randn(n_timesteps)

# Compute the mean and standard deviation over trials:
y_mean = np.mean(y, axis=0)
y_std = np.std(y, axis=0)

# Make the plot:
plt.errorbar(t, y_mean, yerr=y_std, fmt='o')
plt.xlabel('t')
plt.ylabel('y')
````

## Working on our dataset

We will be analyzing data on college majors sourced from the American Community Survey 2010–2012 Public Use Microdata Sample. Similar analysis and more can be found [here](https://realpython.com/pandas-plot-python/). 

Let's start.

````python
#Loading our dataset
import pandas as pd

download_url = (
    "https://raw.githubusercontent.com/fivethirtyeight/"
    "data/master/college-majors/recent-grads.csv"
)

df = pd.read_csv(download_url)
df.head()
````

````python
#Let's create a smaller dataset with just engineering majors. 
#do you remember how to do it from your dataframe chapter?

eng_df = df[df['Major'].str.contains('ENGINEERING')]
eng_df.head()
````

````python
#looking at all column names
list(eng_df.columns)
````

Before we start plotting, let's load our plotting library and set some settings.

````python
import matplotlib.pyplot as plt
%matplotlib inline

plt.rcParams["figure.figsize"] = (20,10)
````

### Bar plots

Bar plots are easy way to visualize data distribution.

````python
#Axis tilt so that we can read them
plt.xticks(rotation = 90)

# general bar graph
plt.bar(eng_df['Major'], eng_df['Men'])
````

Now let's take a look at the ratio between men and women in each degree.

````python
#Now let's plot Men and Women side by side
#Axis tilt so that we can read them
plt.xticks(rotation = 90)

#defining an axis
import numpy as np
x_axis = np.arange(len(eng_df['Major']))

# setting x ticks
plt.xticks(x_axis, eng_df['Major'])

# general bar graph
plt.bar(x_axis -0.2, eng_df['Men'], width=0.4, label="Men")
plt.bar(x_axis +0.2, eng_df['Women'], width=0.4, label="Women")

# Add legend
plt.legend()
````

### Pie chart

Now let's try making a pie chart of the employment for each major. Note that these numbers are not normalized.

````python
#making a pie chart
plt.pie(eng_df['Employed'])

#adding legend
plt.legend(eng_df['Major'], bbox_to_anchor=(1.7,0.9), loc="upper right")
````

Let's normalize this and replot.

````python
#making a pie chart
plt.pie(eng_df['Employed']/eng_df['Total'])

#adding legend
plt.legend(eng_df['Major'], bbox_to_anchor=(1.7,0.9), loc="upper right")
````

### Histogram

Histograms are very useful for showing distributions in data. Often used for data that come from a continuous range or those that can get binned.

Let's plot a histogram of unemployment rates. By default the histogram finds the bins itself, but you can also define the bins.

````python
plt.hist(eng_df['Unemployment_rate'])
````

````python
#Optional
#Sometimes when you're plotting lots of data from a distribution
#you'd like to see that. In this case, you can use another powerful plotting tool
#called seabonr
import seaborn as sns

sns.distplot(eng_df['Unemployment_rate'])
````

### Box plot and violin plot

Box and violin plot both help you see the distribution of your data. Let's try comparing men and women distribution in engineering vs life science fields.

````python
stem_df= df[df['Major_category'].isin(['Computers & Mathematics','Physical Sciences','Biology & Life Science'])]
stem_df.head()
````

Now let's start by plotting Box plot of men distribution in engineering fields.

````python
plt.boxplot([eng_df['Men'], eng_df['Women']], vert=False, labels=['Men','Women'])
````

Now let's add the stem fields side by side.

````python
plt.boxplot([
    eng_df['Men'], eng_df['Women'],
    stem_df['Men'], stem_df['Women']
    ], vert=False, labels=['Men_Eng','Women_eng', 'Men_STEM', 'Women_STEM'])
````

Now let's a look at the violin plot. We will use another plotting package called seaborn for nicer visuals. But matplotlib still works. 

Also, we will use normalzied values for this one.

````python
#let's first generate a new merged dataframe
stem_df_short = stem_df[['Total','Men','Major_category']]
eng_df_short = eng_df[['Total','Men','Major_category']]
frame = [stem_df_short, eng_df_short]

men_df = pd.concat(frame)

men_df['field'] = np.where(men_df['Major_category'] == 'Engineering', 'Eng','Sci_Math')

men_df['gender'] = 'Men'

men_df.columns=['Total','num','Major_category','field', 'gender']

men_df.head()
````

````python
#let's first generate a new merged dataframe
stem_df_short = stem_df[['Total','Women','Major_category']]
eng_df_short = eng_df[['Total','Women','Major_category']]
frame = [stem_df_short, eng_df_short]

women_df = pd.concat(frame)

women_df['field'] = np.where(women_df['Major_category'] == 'Engineering', 'Eng','Sci_Math')

women_df['gender'] = 'Women'

women_df.columns=['Total','num','Major_category','field', 'gender']

women_df.head()
````

````python
merged_df = pd.concat([men_df, women_df])
merged_df.head()
````

````python
#adding normalized column
merged_df['normalized_num'] = merged_df['num']/merged_df['Total']
merged_df.head()
````

````python
import seaborn as sns

#generating a new dataframe with all our data
sns.set(style="whitegrid")
sns.violinplot(x="gender", y="num", hue="field", data=merged_df, 
               palette="muted", split=True)
````

````python
#generating a new dataframe with all our data, normalized
sns.set(style="whitegrid")
sns.violinplot(x="gender", y="normalized_num", hue="field", data=merged_df, 
               palette="muted", split=True)
````
