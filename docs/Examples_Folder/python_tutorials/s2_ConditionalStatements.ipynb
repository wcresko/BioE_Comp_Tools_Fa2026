This exercise walks you through how to check conditions in python.

Your exercise is to write down some code to do the following:

The cell below originates a random number between 1-10. Your job is to try to guess it. Your code should compare your number with the estimated variable, called `mystery_num` and prints statements such as "you got it right", "your number is bigger than the mystery number", and "your number is smaller than the mystery number".

````python
# NOTE: mystery number generation
from random import randint

mystery_num=randint(1,10)
````

## If statement

Python uses `if statement` to check different conditions. The format of the statement is very easy:

```
if condition is true:
  perform action 1
  perform action 2
```

It starts with `if` followed by the condition. The first line *always* ends with **:**. 

It then goes to the next line. Note that the next line is indented. Indentations in python are very important. It means that the second line is nested under the first line and is part of what should happen if the condition is true. If you don't indent the lines, it will not be part of the if statement and will always be performed.

If your condition is false, then none of the actions under if statement will be performed.

HINT: to perform comparisons, you can use `>`, `<`, `==` (means equals, as = is reserved for assigning variables), `!=` (is not equal), `<=` and `>=`.

````python
#your guess
my_guess=4
````

````python
#condition check
if my_guess == mystery_num:
  print("you got it right")
#check condition 2
if my_guess > mystery_num:
  print("your number is bigger than the mystery number")
#check condition 3
if my_guess < mystery_num:
  print("your number is smaller than the mystery number")
````

### if-else statement

When you want to do something if your condition is true and something else if your condition is false, you can use `if-else` statement. The format is:

```
if condition is true:
  perform action 1
else:
  perform action 2
```

Note that `else` is not nested under `if`. Also note that all the indentations should have the same space. Also note that `else` is not followed by any condition. It literally means if condition is false.

**On indentation**: There are generally two ways to add indentations in python. Some people use tabs and some people use spaces. There is a big argument among developers about which to use (`Tabs!` `Two spaces!` `Four spaces!`)  and there are even jokes about it in TV shows that feature tech people! General rule of thumb: be consistent. My preference is two spaces.

````python
if my_guess == mystery_num:
  print("you got it right")
else:
  print("better luck next time")
````

### elif statement

Sometimes, like this case, we have more than one condition to check. In that case, you can use `elif` statement. The format is:

```
if condition1 is true:
  perform action 1
elif condition 2 is true:
  perform action 2
elif condition 3 is true:
  perform action 3
else:
  perform action 4
```

Note that the `elif` lines are not nested under `if`, but follow the same line, and their nested lines follow the same indentation as the one under `if`. You can have an `else` statement at the end, but you don't have to.

Let's try writing our `if` statement using these.

````python
if my_guess == mystery_num:
  print("you got it right")
elif my_guess < mystery_num:
  print("your number is smaller than the mystery number")
else: #you can also use elif, but since your number only can be > now, else works
  print("your number is bigger than the mystery number")
````

**TIP**: To save computational time in complex code, or code that will run on very large datasets, it is recommended to check the most likely condition first and go down based on likelihood. For example, the chance of you guessing the right number is 10%, but you have 45% chance of guessing a number that is smaller and 45% chance of guessing a number that is bigger than the `mystery_num`. So, it is more efficient to put those conditions (`>` and `<`) first. For this code, it wouldn't matter. But it makes a differnce when you're working with big data and difficult calculations.

### Logical operators

You can use logical operators (`and`, `or`, `not`) in your `if` statement to combine conditions and make your code shorter.

````python
#instead of using if and elif or if-else
if my_guess < mystery_num or my_guess > mystery_num:
  print("better luck next time")
````

Let's take a quick refresher on boolean operators. This part is adapted from notebooks by Dr. James Murray, UO.

````python
# For AND, both things being compared must be True for the result to be True:
print(True and True)
print(True and False)
print(False and False)
````

````python
# For OR, at least one of the things must be True for the result to be True:
print(True or True)
print(True or False)
print(False or False or True)
print(False or False)
````
