## For loop

One of the most common ways to repeat a task multiple times. `for loop` is a widely used loop that iterates through...well, anything that is iterable (there are many different iterables in Python - more on this later!). One such example is a range. The format is as follows:

```
for i in range(n1,n2):
  some action
```

At each iteration, the variable in the loop (in this case `i`) is assigned the number we're at. The first number will be `n1`. The last number our variable takes will be `n2-1`. It is important to note that the last number in the range is not considered as part of the range! The variable `i` can be used in the body of the `for loop`.

Let's take a look:

````python
for i in range(1,10):
  print(i)
````

You don't have to use the variable `i` in the body of the loop. It can simply be used to repeat something many times. Look at the example below:

````python
for i in range(1,5):
  print('hello')
````

You can also perform multiple tasks in the body of the loop.

````python
for i in range(1,5):
  n_squared = i**2
  n_cubed = i**3
  print(n_squared, n_cubed)
````

Now let's try to write the code.

````python
for i in range(1,31):
  if i%3 == 0: #a multiple of 3 has mod 3 equal to zero 
               #(that is, a zero remainder when divided by 3)
    print(i)
````

**Looping with a jump**: When we just give two numbers as parameters to the `range` function, it adds to the parameter 1 at a time. You can set the steps in the for loop to be other numbers by passing an additional parameter. 

Let's try.

````python
for i in range(1,10,2):
  print(i)
````

You can also loop in reverse!

````python
for i in range(20,1,-4):
  print(i)
````

Now let's try to write a fancier version of our for loop that does not have an `if statement`.

````python
for i in range(3,31,3):
  print(i)
````

### while loop

Another way to loop in python is to use a `while loop`. While loop is different from the for loop in that it doesn't have an iterable, but it checks a condition. As long as the condition is true, the loop continues. The format is:

```
while condition is true:
  some action
```

Usually the condition is some sort of counter. Let's look at an example:

````python
counter=1
while counter <= 30:
  if counter%3 == 0:
    print(counter)
  counter += 1
````

**Important note**: The line that adds 1 to the counter is very important. If we skip that line, `counter` remains 1 and you will go into a so-called *infinite loop*. So make sure your condition will evaluate to `False` at some point (or `break`, see below).

### Change the flow of a loop

There are two ways to change the flow of a loop:

- `continue`: skips all the lines beneath it in the body of the loop and goes to the next iteration.
- `break`: comes out of the loop completely without finishing the remaining iterations/without the conditional evaluating to `False`.

Let's compare the two below:

````python
for i in range(1,15):
  if i%5 == 0:
    continue
  print(i)
````

````python
for i in range(1,30):
  if i%5 == 0:
    break
  print(i)
````
